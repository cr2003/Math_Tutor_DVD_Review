## 08 Curve Sketching using Derivatives

### Example 1

Given the curve $$y = x^3 + 3x^2 -9x$$

### Useful information:
1. Find x-axis crossing when $y=0$

2. $y'> 0 \implies y$ increasing

3. $y'< 0 \implies y$ decreasing

4. $y' = 0 \implies $ local Max or local Min

5. $y''> 0 \implies y$ is concave upward

6. $y''< 0 \implies y$ is concave downward

7. $y''= 0 \implies $ Inflection Point


```python
x, y = sp.symbols("x, y", real=True)
y = x**3 + 3 * x**2 - 9 * x
dy = sp.diff(y, x)
d2y = sp.diff(y, x, 2).expand()
display(Math(rf" y = {sp.latex(y)}"))
display(Math(rf" y' = {sp.latex(dy)}"))
display(Math(rf" y'' = {sp.latex(d2y)}"))
```

```python
# Create figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# Call the helper function (grid is enabled by default)
setup_math_axes(ax, xlim=(-6, 6), ylim=(-25, 150))
x = np.arange(-5, 5, 0.1)
y = x**3 + 3 * x**2 - 9 * x
dy = 3 * x**2 + 6 * x - 9
d2y = 6 * x + 6
plt.plot(x, y, label="$y = x^3 + 3x^2 -9x$")
plt.plot(x, dy, label="$y' =  3x^2 + 6x -9$")
plt.plot(x, d2y, label="$y'' = 6x + 6$")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

### Useful information:
1. Find x-axis crossing when $y=0$

2. $y'> 0 \implies y$ increasing

3. $y'< 0 \implies y$ decreasing

4. $y' = 0 \implies $ local Max or local Min

5. $y''> 0 \implies y$ is concave upward

6. $y''< 0 \implies y$ is concave downward

7. $y''= 0 \implies $ Inflection Point


1. Finding x-axis crossings y=0

```python
x_crossing = sp.solve(y, x)

# Apply evalf() to each element in the list
x_crossing_float = [val.evalf() for val in x_crossing]

display(
    Math(
        rf"x\text{{-crossing points}} = {sp.latex(x_crossing)} = {sp.latex(x_crossing_float)}"
    )
)
```

2. Take derivative and find local max or min, f'(x)=0

```python
local_pts = sp.solve(dy)
display(Math(rf"\text{{local max, min}} = {sp.latex(local_pts)}"))
```

### Example 2

Given the equation $f(x) = x^3 - x$ find the interesting points of the curve.

```python
(
    fx,
    x,
) = sp.symbols("f, x", real=True)

fx = x**3 - x
dfx = sp.diff(fx, x)
d2fx = sp.diff(fx, x, 2).expand()
display(Math(rf" f(x) = {sp.latex(fx)}"))
display(Math(rf" f'(x) = {sp.latex(dfx)}"))
display(Math(rf" f''(x) = {sp.latex(d2fx)}"))
```

Finding the x-crossing points y=0

```python
x_crossing = sp.solve(fx, x)

display(Math(rf"x\text{{-crossing points}} = {sp.latex(x_crossing)}"))
```

Finding focal max and local min

```python
local_pts = sp.solve(dfx)
# Apply evalf() to each element in the list
xx = [val.evalf() for val in local_pts]
display(Math(rf"\text{{x-max, x-min}} = {sp.latex(xx)}"))
yy = [fx.subs(x, pt) for pt in xx]
display(Math(rf"\text{{y-max, y-min}} = {sp.latex(yy)}"))
```

Findind inflection point when y''=0

```python
inf_pt_x = sp.solve(d2fx, x)[0]
inf_pt_y = fx.subs(x, inf_pt_x)
display(
    Math(rf"\text{{inflection Piont}} = ({sp.latex(inf_pt_x)}, {sp.latex(inf_pt_y)})")
)
```

```python
# Create figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# Call the helper function (grid is enabled by default)
setup_math_axes(ax, xlim=(-1.5, 1.5), ylim=(-0.5, 0.5))

x = np.arange(-1, 1.1, 0.01)
y = x**3 - x

plt.plot(x, y, label="$y = x^3 -x$")
# add the inflection point [0,0]) and points of interest (local max, min)
ptx = [0] + xx
pty = [0] + yy
plt.plot(ptx, pty, "ro")

plt.legend()

plt.show()
```
