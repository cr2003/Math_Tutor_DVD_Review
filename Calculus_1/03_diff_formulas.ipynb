## Differentiation Formulas

### Derivatives
An unevaluated derivative is created by using the `Derivative`. To evaluate an unevaluated derivative, use the doit method.

There is not an easy way to define a constant in sympy, so remember that the derivative of a constant is zero:

$$ \frac{d}{d x} C = 0$$

$$ \frac{d}{d x} x = 1$$


```python
expr = x
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

$$ \frac{d}{d x} C f(x)= C \frac{df}{d x}$$


```python
expr = 5 * x
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

$$\frac{d}{d x}(x^n)=nx^{n-1}$$

```python
expr = x**2
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = x ** (-3)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = 2 * x**5
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = 3 * x ** (-4)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

$$(f+g)'=f' + g'$$

$$(f-g)'=f'- g'$$



```python
expr = x**2 - 10 * x + 100
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

Volume

$$V(r)=\frac{4}{3}\pi r^3 \implies V'(r)=\frac{4}{\cancel{3}}\pi \cancel{3} r^2=4 \pi r^2$$


```python
expr = sp.Rational(4, 3) * sp.pi * r**3
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```


$$(f g)' =f' g + f g' $$

```python
expr = (x + 2) / x
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

$$\left(\frac{f}{g} \right)' = \frac{gf'-fg'}{g^2}$$

```python
expr = x**5 / (x**3 - 2)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = x + (x**2) ** sp.Rational(1, 5)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```
