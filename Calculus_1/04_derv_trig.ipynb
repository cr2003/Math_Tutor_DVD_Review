## Derivatives of Trigonometric Functions

### Derivatives
An unevaluated derivative is created by using the `Derivative`. To evaluate an unevaluated derivative, use the doit method.

```python
expr = sp.sin(x)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = sp.cos(x)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = sp.tan(x)
d = sp.Derivative(expr)
sp.Eq(d, sp.trigsimp(d.doit()).rewrite("sec"))
```

```python
expr = sp.cot(x)
d = sp.Derivative(expr)
sp.Eq(d, sp.trigsimp(d.doit()).rewrite("csc"))
```

```python
expr = sp.sec(x)
d = sp.Derivative(expr)
sp.Eq(d, sp.trigsimp(d.doit()))
```

```python
expr = sp.csc(x)
d = sp.Derivative(expr)
sp.Eq(d, sp.trigsimp(d.doit()))
```

```python
expr = sp.sin(x) + sp.cos(x)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

```python
expr = x / (sp.sin(x) + sp.cos(x))
d = sp.Derivative(expr)
# sp.Eq(d, sp.trigsimp(d.doit()))
sp.Eq(d, d.doit())
```
