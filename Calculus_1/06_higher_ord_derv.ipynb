## Higher Order Derivatives

Given a function  $y$, we can take a chain of derivatives of the function.

So we can take the first derivative

$$\frac{d}{dx}y \quad, \quad y'$$

and then the second derivative: 

$$\frac{d^2}{dx^2}y \quad, \quad y''$$

and then the third derivative: 

$$\frac{d^3}{dx^3}y \quad, \quad y''' , \quad y^{(3)}$$


### Example 1
Given the following function, find the first and second derivatives.
$$f(x) = x^4 -3x^3 + 16x$$

```python
import pprint

expr = x**4 - 3 * x**3 + 16 * x
d = sp.Derivative(expr, x)
d2 = sp.Derivative(expr, x, 2)

print("First derivative:")
sp.Eq(d, d.doit())
```

```python
print("Second derivative:")
sp.Eq(d2, d2.doit())
```

### Example 2
Given the following function, find the second derivative.
$$h(x) = \sqrt{x^2 + 1} \quad \quad h''(x) = \frac{-x^2}{(x^2+1)^{3/2}} + \frac{1}{(x^2+1)^{1/2}}$$

```python
expr = (x**2 + 1) ** sp.Rational(1, 2)
d = sp.Derivative(expr, x)
sp.Eq(d, d.doit())
```

```python
d = sp.Derivative(expr, x, 2)
# sp.Eq(d, d.doit())
sp.Eq(d, sp.simplify(d.doit()))
```
