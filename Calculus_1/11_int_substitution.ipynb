## 10 Integration by Substitution
The idea is to transform some function over $x$ to another function over lets say $u$. For instance if we have

$$\int x(x^2-1)^{99} \, dx$$

we can transform $$"x" \implies "u"$$  

$$u = x^2 -1$$

then

$$\frac{du}{dx}= 2x$$
solving for $du$

$$du = 2x dx\quad; \quad dx = \frac{1}{2x} du $$

now we substitute in the original integral

$$\int x u^{99} \, \frac{1}{2x} du$$

$$\int  \frac{1}{2} u^{99} \, du$$

Now we solve as a standard integral

$$\int  \frac{1}{2} u^{99} \, du = \frac{1}{2} \frac{1}{100}u^{100} =   \frac{1}{200}u^{100} + C$$

and in the final answer we substitute back the terms in $u$ by $u = x^2 -1$ 

$$\frac{1}{200}u^{100} + C = \frac{1}{200}(x^2 -1)^{100} + C$$

```python
expr = x * (x**2 - 1) ** 99
result = sp.integrate(expr, x)
simplified_result = sp.factor(result)

display(sp.Eq(sp.Integral(expr, x), simplified_result))
```
