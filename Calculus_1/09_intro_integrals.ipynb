## 09 Introduction to Integrals
### Integrals

If we have the function $$f(x) = x^2$$

and we take its __derivative__ 

$$f'(x) = 2x$$

we can do the opposite, that is take the __antiderivative__ or __Integral__ 

$$\int 2x \, dx = x^2 + C$$





```python
expr = x**2
sp.Eq(f(x), expr)
```

```python
expr_der = expr.diff()
sp.Eq(sp.Derivative(f(x)), expr_der)
```

```python
sp.Eq(sp.Integral(expr_der), expr_der.integrate() + C)
```

### Integral Example

```python
expr = 3 * x**2
sp.Eq(sp.Integral(expr), sp.integrate(expr) + C)
```

### Integral Properties
#### 1. Polynomials

```python
C, n = sp.symbols("C, n", positive=True)
r, x, y, z = sp.symbols("r, x, y, z")
f, g, h = sp.symbols("f, g, h", cls=sp.Function)
```

```python
expr = x**n
sp.Eq(sp.Integral(x**n, x), sp.integrate(expr, x))
```

```python
expr = x**57
sp.Eq(sp.Integral(x**n, x), sp.integrate(expr, x))
```

#### 2. Constants

```python
expr = C * f(x)
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

```python
expr = 4 * x**5
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

#### 3. Add functions

```python
expr1 = f(x)
expr2 = g(x)
sp.Eq(sp.Integral(expr1 + expr2, x), sp.integrate(expr1, x) + sp.integrate(expr2, x))
```

```python
expr = x**3 + 4 * x**4
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

#### 4. Trig functions

```python
expr = sp.sin(x)
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

```python
expr = sp.cos(x)
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

```python
expr = sp.sec(x) ** 2
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

```python
expr = sp.csc(x) ** 2
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

```python
expr = sp.sec(x) * sp.tan(x)
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

```python
expr = sp.csc(x) * sp.cot(x)
sp.Eq(sp.Integral(expr, x), sp.integrate(expr, x))
```

### Meaning of an Integral
What does an integral mean? $\implies$ Area under the curve

If we have the function $y = x^2$ and we want to calculate the area between 0 and 4 we have the definite integral:

$$\text{Area} = \int_0^4 x^2 \, dx$$

```python
expr = x**2
sp.Eq(sp.Integral(expr, (x, 0, 4)), sp.integrate(expr, (x, 0, 4)))
```
