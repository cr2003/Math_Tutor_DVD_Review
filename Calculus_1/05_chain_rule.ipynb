## The Chain Rule

Given two functions $f(x)$ and $g(x)$, if we have a composite function $F(x) = f(g(x))$, to take the derivative of the composite function $F(x)$ we apply the chain rule:

$$F(x) = f(g(x))$$

$$F'(x) = f'(g(x)) \cdot g'(x)$$

Like in the following example:

### Example 1
$$y = \sin(x^2) \quad\quad y' = 2x \cos(x^2)$$

```python
expr = sp.sin(x**2)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

### Example 2

$$g(x) = \sqrt{x^2-7x} \quad\quad g'(x) = \frac{2x-7}{2\sqrt{x^2-7}}$$

```python
expr = (x**2 - 7 * x) ** sp.Rational(1, 2)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

### Example 3

$$y = \tan(3x) \quad\quad y' = 3\sec^2(3x)$$

```python
expr = sp.tan(3 * x)
d = sp.Derivative(expr)
# sp.Eq(d, d.doit())
sp.Eq(d, sp.trigsimp(d.doit()).rewrite("sec"))
```

### Example 4

$$y = (1+\cos^2x)^6 \quad\quad y' = -12(1+\cos^2x)^5 (\cos x \sin x)$$

```python
expr = (1 + (sp.cos(x) ** 2)) ** 6
d = sp.Derivative(expr)
# sp.Eq(d, sp.trigsimp(d.doit()))
sp.Eq(d, d.doit())
```

### Example 5

$$y = \sin^3 x + \cos^3 x\quad\quad y' = 3\sin x \cos x(\sin x - \cos x )$$

```python
expr = sp.sin(x) ** 3 + sp.cos(x) ** 3
d = sp.Derivative(expr)
# sp.Eq(d, sp.trigsimp(d.doit()))
sp.Eq(d, d.doit())
```

### Example 6

$$y = \tan^2(x^3) \quad\quad y' = 6x^2 \tan(x^3) \sec^2(x^3)$$

```python
expr = sp.tan(x**3) ** 2
d = sp.Derivative(expr)
sp.Eq(d, sp.trigsimp(d.doit()))
# sp.Eq(d, d.doit())
```

### Example 7

$$f(x) = \left(\frac{x-6}{x+7}\right)^3 \quad\quad f'(x) = \frac{39(x-6)^2}{(x+7)^4}$$

```python
expr = ((x - 6) / (x + 7)) ** 3
d = sp.Derivative(expr)
# sp.Eq(d, sp.trigsimp(d.doit()))
# sp.Eq(d, d.doit())
sp.Eq(d, sp.simplify(d.doit()))
```
