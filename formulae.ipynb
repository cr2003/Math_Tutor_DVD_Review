## Differentiation Formulas

### Derivatives
An unevaluated derivative is created by using the `Derivative`. To evaluate an unevaluated derivative, use the doit method.

There is not an easy way to define a constant in sympy, so remember that the derivative of a constant is zero:

$$ \frac{d}{d x} C = 0$$

$$ \frac{d}{d x} x = 1$$


$$ \frac{d}{d x} C f(x)= C \frac{df}{d x}$$


$$\frac{d}{d x}(x^n)=nx^{n-1}$$

$$(f+g)'=f' + g'$$

$$(f-g)'=f'- g'$$

$$V(r)=\frac{4}{3}\pi r^3 \implies V'(r)=\frac{4}{\cancel{3}}\pi \cancel{3} r^2=4 \pi r^2$$



$$(f g)' =f' g + f g' $$

$$\left(\frac{f}{g} \right)' = \frac{gf'-fg'}{g^2}$$

## The Chain Rule

Given two functions $f(x)$ and $g(x)$, if we have a composite function $F(x) = f(g(x))$, to take the derivative of the composite function $F(x)$ we apply the chain rule:

$$F(x) = f(g(x))$$

$$F'(x) = f'(g(x)) \cdot g'(x)$$
