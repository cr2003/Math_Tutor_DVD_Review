## The Chain Rule

Given two functions $f(x)$ and $g(x)$, if we have a composite function $F(x) = f(g(x))$, to take the derivative of the composite function $F(x)$ we apply the chain rule:

$$F(x) = f(g(x))$$

$$F'(x) = f'(g(x)) \cdot g'(x)$$

#### If we want instead around the y-axis

To find the volume of the solid generated by rotating the region bounded by $f(x) = \sqrt{x}$ and the interval from $x = 0$ to $x = 1$ about the y-axis, we can use the shell method (or the washer method with integration with respect to $y$).

**Shell Method Setup:**

Using cylindrical shells, the radius of a shell at position $x$ is $x$, the height is $f(x) = \sqrt{x}$, and the thickness is $dx$.The volume integral is:

$$V = \int_{a}^{b} 2\pi x f(x) \, dx = \int_{0}^{1} 2\pi x \sqrt{x} \, dx = 2\pi \int_{0}^{1} x^{3/2} \, dx$$

## Higher Order Derivatives

Given a function  $y$, we can take a chain of derivatives of the function.

So we can take the first derivative

$$\frac{d}{dx}y \quad, \quad y'$$

and then the second derivative: 

$$\frac{d^2}{dx^2}y \quad, \quad y''$$

and then the third derivative: 

$$\frac{d^3}{dx^3}y \quad, \quad y''' , \quad y^{(3)}$$


## Differentitation of an imiplicit equation

Implicit differentiation is when we don't have a explicit equality  of the variable, (solved by y), such as:

$$3y^2 = -2x^4$$

then we differentiate both sides of the equation: 

$$6y \frac{dy}{dx} = -8x^3$$

and solved by de $\frac{dy}{dx}$: 

$$\frac{dy}{dx} = \frac{-4x^3}{3y}$$
