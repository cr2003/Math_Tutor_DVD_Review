#### Display:


```python
# Define the symbolic time variable and constants
t = sp.Symbol("t")
b = sp.Symbol("b")
c = sp.Symbol("c")

# Define position function x(t)
x = b * t**2 - c * t**3

# Compute velocity v(t) as the derivative of x(t) with respect to t
v = sp.diff(x, t)

# Display the resulting expression
display(Math(rf"v = {sp.latex(v)}"))
```
