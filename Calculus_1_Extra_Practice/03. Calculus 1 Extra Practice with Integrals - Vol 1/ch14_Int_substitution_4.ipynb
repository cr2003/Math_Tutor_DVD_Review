## Video 14. Integration by Substitution, Part 4


```python
f = sp.Integral((3 * x**2) * (x**3 + 1) ** 12, x)
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)} + C"))
```

```python
# check answer with the video
Video_answer = sp.Rational(1, 13) * (x**3 + 1) ** 13
display(Math(rf"\text{{Video answer = }} {sp.latex(Video_answer)} + C"))

# Check if both have the exact same derivative (0 difference)
difference = sp.simplify(sp.diff(answer - Video_answer, x))  # Returns 0
print(f"Check the difference (answer - Video_answer) = {difference}")
```

```python
f = sp.Integral(sp.sin(sp.pi * x), (x, -1, 3))
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)}"))
```

#### Check graphically the answer

```python
# Sample x values across [-1, 3]
x_vals = np.linspace(-1.0, 3.0, 1000)
y_vals = np.sin(np.pi * x_vals)

fig, ax = plt.subplots(figsize=(8, 4.5))

# Plot the curve
ax.plot(x_vals, y_vals, color="tab:blue", linewidth=2, label=r"$y = \sin(\pi x)$")

# Shade positive signed area (y >= 0) and negative signed area (y < 0)
ax.fill_between(
    x_vals,
    y_vals,
    0,
    where=(y_vals >= 0),
    color="tab:green",
    alpha=0.25,
    interpolate=True,
    label="Positive area (+)",
)
ax.fill_between(
    x_vals,
    y_vals,
    0,
    where=(y_vals < 0),
    color="tab:red",
    alpha=0.25,
    interpolate=True,
    label="Negative area (−)",
)

# Horizontal zero line and axis styling
ax.axhline(0, color="black", linewidth=1)
ax.set_xlim(-1.0, 3.0)
ax.set_ylim(-1.25, 1.25)
ax.set_xticks([-1, 0, 1, 2, 3])
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$\sin(\pi x)$")
ax.set_title(
    r"Graph of $\sin(\pi x)$ on $[-1, 3]$  ($\int_{-1}^{3}\sin(\pi x)\,dx = 0$)"
)
ax.grid(True, linestyle="--", alpha=0.4)
ax.legend(loc="lower right")

plt.tight_layout()
plt.show()
```

#### Check graphically the answer

```python
# Sample x values across [0, pi/4]
x_vals = np.linspace(0, np.pi / 4, 1000)
y_vals = np.sin(x_vals) / (np.cos(x_vals) ** 2)

fig, ax = plt.subplots(figsize=(8, 4.5))

# Plot the curve
ax.plot(
    x_vals, y_vals, color="tab:blue", linewidth=2, label=r"$\frac{\sin x}{\cos^2 x}$"
)

# Shade positive signed area (y >= 0) and negative signed area (y < 0)
ax.fill_between(
    x_vals,
    y_vals,
    0,
    where=(y_vals >= 0),
    color="tab:green",
    alpha=0.25,
    interpolate=True,
    label="Positive area (+)",
)
ax.fill_between(
    x_vals,
    y_vals,
    0,
    where=(y_vals < 0),
    color="tab:red",
    alpha=0.25,
    interpolate=True,
    label="Negative area (−)",
)

# Horizontal zero line and axis styling
ax.axhline(0, color="black", linewidth=1)
ax.set_xlim(0.0, np.pi / 4)
ax.set_ylim(-1, 2)

# ax.set_xticks in pi fractions
tick_pos = [
    0,
    np.pi / 16,
    np.pi / 8,
    3 * np.pi / 16,
    np.pi / 4,
]
tick_labels = [
    r"$0$",
    r"$\frac{\pi}{16}$",
    r"$\frac{\pi}{8}$",
    r"$\frac{3\pi}{16}$",
    r"$\frac{\pi}{4}$",
]

ax.set_xticks(tick_pos)
ax.set_xticklabels(tick_labels)

ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$\frac{\sin x}{\cos^2 x}$")

ax.set_title(
    r"Graph of $\frac{\sin x}{\cos^2 x}$ on $[0, \pi/4]$  ($\int_{0}^{\pi/4}\sin(x)/\cos^2(x)\,dx = \sqrt{2}- 1$)"
)
ax.grid(True, linestyle="--", alpha=0.4)
ax.legend(loc="lower right")

plt.tight_layout()
plt.show()
```
