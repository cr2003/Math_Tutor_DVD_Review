## Video 19. Integrals of Exponential Functions, Part 2


### Exercise 1
$$\int \frac{e^{t}}{e^{t} + 1}\, dt$$

```python
f = sp.Integral(sp.E**t / (sp.E**t + 1), (t))
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)} + C"))
```

### Exercise 2
$$\int \frac{e^{2t}} {\sqrt{e^{2t} - 4}}\, dt$$

```python
f = sp.Integral(sp.E ** (2 * t) / sp.sqrt((sp.E ** (2 * t) - 4), (t)))
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)} + C"))
```
