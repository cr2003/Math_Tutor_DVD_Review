## Video 15. The Logarithm as an Integral, Part 1


$$\int \frac{1}{x} dx = \ln |x| + C$$

$$\frac{d}{dx} \ln x  = \frac{1}{x}$$


<span style='color:green'> &#128393; __Note:__ </span>

<div class="alert alert-block alert-success">
    
Remember than in SymPy log = ln, and ln of negative numbers are complex numbers, in reals are not defined.

</div>


```python
f = sp.Integral(1 / x, x)
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)} + C"))
```

```python
# example of log of a negative number as a complex answer
sp.log(-3)
```

```python
f = sp.Integral(1 / x, (x, 1, 3))
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)}"))
```

```python
f = sp.Integral(-1 / (3 * x), (x, sp.Rational(1, 9), sp.Rational(1, 4)))
answer = f.doit()
display(Math(rf"{sp.latex(f)} = {sp.latex(answer)}"))
```

```python
# check answer with the video
Video_answer = sp.Rational(1, 3) * sp.log(sp.Rational(4, 9))
display(Math(rf"\text{{Video answer = }} {sp.latex(Video_answer)}"))

# Check if both have the exact same derivative (0 difference)
difference = sp.simplify(sp.diff(answer - Video_answer, x))  # Returns 0
print(f"Check the difference (answer - Video_answer) = {difference}")
```
