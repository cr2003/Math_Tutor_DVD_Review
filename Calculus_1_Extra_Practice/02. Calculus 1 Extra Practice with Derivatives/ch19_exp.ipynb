```python
expr = sp.exp(x)
sp.Eq(sp.Derivative(expr), expr.diff())
```

### Example 1

```python
expr = sp.exp(4 * x)
sp.Eq(sp.Derivative(expr), expr.diff())
```

### Example 2

```python
expr = sp.exp(x**5)
sp.Eq(sp.Derivative(expr), expr.diff())
```

### Example 3

```python
expr = (sp.exp(x) + 1) / (sp.exp(x) - 1)
sp.Eq(sp.Derivative(expr), expr.diff().factor())
```

### Example 4

```python
expr = sp.tan(sp.exp(3 * z))
sp.Eq(sp.Derivative(expr), expr.diff().simplify().rewrite("sec"))
```

### Example 5

```python
expr = (x + 1) * sp.exp(-x)
sp.Eq(sp.Derivative(expr), expr.diff().factor())
```
