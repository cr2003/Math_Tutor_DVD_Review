## The Chain Rule

Given two functions $f(x)$ and $g(x)$, if we have a composite function $F(x) = f(g(x))$, to take the derivative of the composite function $F(x)$ we apply the chain rule:

$$F'(x) = f'(g(x)) \cdot g'(x)$$

Like in the following example:



### Example 1
$$y = \sin(x^2) \quad\quad y' = \cos(x^2)\, 2x =  2x\, \cos(x^2)$$

```python
expr = sp.sin(x**2)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

### Example 2
$$f(x) = \left(4-3x^2\right)^{400}$$

```python
expr = (4 - 3 * x**2) ** 400
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

### Example 3
$$f(x) = \sin(5x)$$

```python
expr = sp.sin(5 * x)
d = sp.Derivative(expr)
sp.Eq(d, d.doit())
```

### Example 4
$$f(x) = \tan^6(x)$$

```python
expr = (sp.tan(x)) ** 6
d = sp.Derivative(expr)
sp.Eq(d, d.doit().simplify())
```

### Example 5
$$f(x) = \sqrt[3]{1-\sin x}$$

```python
expr = sp.root(1 - sp.sin(x), 3)
d = sp.Derivative(expr)
sp.Eq(d, d.doit().simplify())
```

### Example 6
$$f(x) =\cos(\sin (t))$$

```python
expr = sp.cos(sp.sin(t))
d = sp.Derivative(expr)
sp.Eq(d, d.doit().simplify())
```

### Example 7
$$f(x) = \left(\frac{x-1}{x+1} \right)^3$$


```python
expr = ((x - 1) / (x + 1)) ** 3
d = sp.Derivative(expr)
sp.Eq(d, d.doit().simplify())
```
