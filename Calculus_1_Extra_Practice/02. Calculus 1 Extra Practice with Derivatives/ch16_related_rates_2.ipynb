```python
# Define symbolic variables
t = sp.Symbol("t", real=True)
h = sp.Function("h")(t)
dhdt = sp.Symbol("dhdt", real=True)
dVdt = sp.Symbol("dVdt", real=True)

# Cone dimensions
H = sp.symbols("H", positve=True)
R = sp.symbols("R", positve=True)


# Radius as a function of height h using similar triangles: r = (R/H)*h
r = (R / H) * h

# Volume of water in the cone: V = (1/3) * pi * r^2 * h
V = (sp.Rational(3)) * sp.pi * r**2 * h

# Differentiate volume with respect to time t
# Substitute the derivative of h(t) with dhdt for algebraic handling
diff_V = V.diff(t).subs(sp.Derivative(h, t), dhdt)

display(Math(rf"V = {sp.latex(V)}"))
display(Math(rf"\frac{{dV}}{{dt}} = {sp.latex(diff_V)}"))
```

```python
# Define AstroPy Quantity variables using the '_qty' suffix
h_qty = 2.0 * u.m
H_qty = 6.0 * u.m
R_qty = 2.0 * u.m
dhdt_qty = 0.5 * (u.m / u.s)

# Create the numerical function using lambdify from the sympy expression
# diff_V depends on h and dhdt
f_dVdt = sp.lambdify((R, dhdt, h, H), diff_V, "numpy")

# Evaluate directly using AstroPy Quantity objects
dVdt_qty = f_dVdt(R_qty, dhdt_qty, h_qty, H_qty)

display(Math(rf"\text{{Volume inflow rate: }} {dVdt_qty:.4f}"))

display(
    Math(rf"\left.\frac{{dV}}{{dt}}\right|_{{h=2\text{{ m}}}} \approx {dVdt_qty:.4f}")
)
```

```python
# Define time t, angle theta as a function of t, and angular velocity dthetadt
t = sp.Symbol("t", real=True)
theta = sp.Function("theta")(t)
dthetadt = sp.Symbol("dthetadt", real=True)

# Distance along the shore x as a function of theta
# x = tan(theta) since the distance to shore is 1 km
x = sp.tan(theta)

# Differentiate x with respect to time t implicitly
# Substitute the derivative of theta(t) with dthetadt
dxdt_expr = x.diff(t).subs(sp.Derivative(theta, t), dthetadt)

display(Math(rf"x = {sp.latex(x)}"))
display(Math(rf"\frac{{dx}}{{dt}} = {sp.latex(dxdt_expr)}"))
```

```python
# Define AstroPy Quantity variables for the given values (_qty suffix)
d_qty = 1.0 * u.km  # Distance from lighthouse to shore
s_qty = 2.0 * u.km  # Distance from lighthouse to the beam spot on sand
dthetadt_qty = 10.0 * np.pi * (u.rad / u.min)

# From geometry, cos(theta) = d / s, so sec(theta) = s / d
# Therefore, sec^2(theta) = (s / d)**2
sec_theta_sq = (s_qty / d_qty) ** 2

# Lambdify the symbolic expression (dxdt_expr contains sp.sec(theta)**2 * dthetadt)
# Note: sp.tan(theta).diff(t) results in sp.sec(theta)**2 * Derivative(theta, t)
# f_dxdt = lambdify((theta, dthetadt), dxdt_expr, 'numpy')

# Since dxdt_expr uses sp.sec(theta), we can evaluate it using numpy/astropy directly:
# sec(theta)**2 * dthetadt
dxdt_qty = sec_theta_sq * dthetadt_qty * u.km

print(f"Speed of the beam along the shore: {dxdt_qty:.4f}")
display(Math(rf"\frac{{dx}}{{dt}} \approx {dxdt_qty.value:.4f} \, \text{{km/min}}"))
```

```python
# Define independent variable t, ramp position function s(t), and speed dsdt
t = sp.Symbol("t", real=True)
s = sp.Function("s")(t)
dsdt = sp.Symbol("dsdt", real=True)

# Incline dimensions
H = 5
L = 20

# Height as a function of ramp distance s using similar triangles
# h = (H / L) * s
h = sp.Rational(H, L) * s


# Differentiate height with respect to time t
diff_h = h.diff(t).subs(sp.Derivative(s, t), dsdt)

display(Math(rf"h = {sp.latex(h)}"))
display(Math(rf"\frac{{dh}}{{dt}} = {sp.latex(diff_h)}"))
```

```python
# Define AstroPy Quantity variables using the '_qty' suffix
dsdt_qty = 3.0 * (u.m / u.s)

# Create the numerical function using lambdify from the sympy expression
f_dhdt = sp.lambdify((dsdt,), diff_h, "numpy")

# Evaluate directly using AstroPy Quantity objects
dhdt_qty = f_dhdt(dsdt_qty)

print(f"Box rise speed: {dhdt_qty:.4f}")
display(Math(rf"\frac{{dh}}{{dt}} \approx \, {dhdt_qty:.4f}"))
```
