## Video 15 - Related Rates - Part 1

### Example 1
The radius of a spherical balloon shrinks at 1/2 meter per minute. How fast is the volume decresing when the radius is 4 meters?

#### Solution:


```python
d, dVdt, dVdr, drdt, r, x, y, z = sp.symbols(
    "d, dVdt, dVdr, drdt, r, x, y, z", real=True
)

expr = sp.Rational(4, 3) * sp.pi * r**3
der = sp.Derivative(expr)
sol = der.doit()
eq = sp.Eq(der, sol)
display(Math(rf"V = {sp.latex(expr)}"))
display(Math(rf"\frac{{dV}}{{dr}} = {sp.latex(eq)}"))
eq1 = sp.Eq(dVdt, dVdr * drdt)
display(
    Math(
        rf"\frac{{dV}}{{dt}} = \frac{{dV}}{{dr}} \frac{{dr}}{{dt}} = {sp.latex(sol)} \frac{{dr}}{{dt}} "
    )
)


resp = eq1.subs({dVdr: sol, drdt: -1 / 2, r: 4})
display(Math(rf"\frac{{dV}}{{dt}} =   {sp.latex(resp)} \; {{m^3/min}}"))
```

### Example 2
Water leaks onto a floor and creates a circular pool with area that increases at a rate of 3 square meters per minute. How fast is the radius of the pool increasing when the radius is 10 meters?

#### Solution:


```python
d, dAdt, dAdr, drdt, r, x, y, z = sp.symbols(
    "d, dAdt, dAdr, drdt, r, x, y, z", real=True
)
A = sp.pi * r**2
der = sp.Derivative(A)
sol = der.doit()
eq1 = sp.Eq(der, sol)
eq2 = sp.Eq(dAdt, dAdr * drdt)
resp = eq2.subs({dAdt: 3, dAdr: sol, r: 10})
solution = sp.solve(resp, drdt)[0]

# --- print results ---
display(Math(rf"A = {sp.latex(A)}"))
display(Math(rf"\frac{{dA}}{{dr}} = {sp.latex(sol)}"))
display(
    Math(
        rf"\frac{{dA}}{{dt}} = \frac{{dA}}{{dr}} \frac{{dr}}{{dt}} = {sp.latex(sol)} \frac{{dr}}{{dt}} "
    )
)
display(Math(rf"\frac{{dr}}{{dt}} =   {sp.latex(solution)} \; {{m/min}}"))
```

### Example 3
A ladder of length of 13 meters. Suppose the top of the ladder is pushed up a wall at 1 meter per second. How fast is the base of the ladder approaching the wall when it is 3 meters from the wall?


#### Solution:
<img src="./pics/Fig15_1.png" width=400> 


```python
# Define independent variable t and constant L
t = sp.Symbol("t", real=True)
L = sp.Symbol("L", real=True, positive=True)

# Define x and y as functions of t
x = sp.Function("x")(t)
y = sp.Function("y")(t)

# Define the equation x^2 + y^2 = L^2
eq = sp.Eq(x**2 + y**2, L**2)

# Differentiate both sides of the equation with respect to t
diff_eq = sp.Eq(eq.lhs.diff(t), eq.rhs.diff(t))

# Display the results
display(Math(rf"\text{{Original Equation: }} {sp.latex(eq)}"))
display(Math(rf"\text{{Differentiated w.r.t. }} t: {sp.latex(diff_eq)}"))
```

```python
# Define symbols for the derivatives so we can isolate them easily
dxdt = sp.Symbol("dxdt", real=True)
dydt = sp.Symbol("dydt", real=True)

# Substitute the Derivative function objects with plain symbols for solving
diff_expr = diff_eq.lhs.subs({sp.Derivative(x, t): dxdt, sp.Derivative(y, t): dydt})

# Solve for dxdt algebraically
dxdt_solution = sp.solve(sp.Eq(diff_expr, 0), dxdt)[0]

display(Math(rf"\frac{{dx}}{{dt}} = {sp.latex(dxdt_solution)}"))
```

```python
# 1. Define AstroPy quantities (_qty suffix)
L_qty = 13.0 * u.m
x_qty = 3.0 * u.m
dydt_qty = 1.0 * (u.m / u.s)
y_qty = np.sqrt(L_qty**2 - x_qty**2)

# 2. Create the numerical function using lambdify
# Map the free symbols in dxdt_solution to the function arguments
f_dxdt = sp.lambdify((x, y, L, dydt), dxdt_solution, "numpy")

# 3. Evaluate directly with AstroPy quantities
# Note: x and y in the lambdify arguments match x(t) and y(t) symbols from Cell 1
dxdt_result_qty = f_dxdt(x_qty, y_qty, L_qty, dydt_qty)

print(f"Rate: {dxdt_result_qty:.4f}")
```
