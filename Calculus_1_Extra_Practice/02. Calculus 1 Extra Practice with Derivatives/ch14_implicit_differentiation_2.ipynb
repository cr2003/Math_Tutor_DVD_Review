## Video 14 - Implicit Differentiation - Part 2


### Example 1
Given the following equation 

$$x^2 + y^2 = \frac{y^2}{x^2}$$

find $\frac{dy}{dx}$ using implicit differentiation.


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(x**2 + y**2, y**2 / x**2)
expr = x**2 + y**2 - y**2 / x**2
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
```

### Example 2 
Given the following equation 

$$x^2 + y^2 = y$$

find $\frac{dy}{dx}$ at point (0,1)


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(x**2 + y**2, y)
expr = x**2 + y**2 - y
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
sol = dy_dx.subs({x: 0, y: 1})
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
display(Math(rf"{sp.latex(d)}@(0,1) = {sp.latex(sol)}"))
```

### Example 3
Given the following equation 

$$x^3 + 2xy = 5$$

find $\frac{dy}{dx}$ at point (1,2):


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(x**3 + 2 * x * y, 5)
expr = x**3 + 2 * x * y - 5
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
sol = dy_dx.subs({x: 1, y: 2})
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
display(Math(rf"{sp.latex(d)}@(1,2) = {sp.latex(sol)}"))
```

### Example 4 
Given the following equation 

$$(\sqrt{x}+1)(\sqrt{y}+2)= 8$$

find $\frac{dy}{dx}$ at point (1,4)


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq((sp.sqrt(x) + 1) * (sp.sqrt(y) + 2), 8)
expr = (sp.sqrt(x) + 1) * (sp.sqrt(y) + 2) - 8
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
sol = dy_dx.subs({x: 1, y: 4})
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
display(Math(rf"{sp.latex(d)}@(1,4) = {sp.latex(sol)}"))
```
