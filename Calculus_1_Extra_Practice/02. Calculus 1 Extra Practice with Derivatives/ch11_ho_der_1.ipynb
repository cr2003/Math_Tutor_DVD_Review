## Higher Order Derivatives

Given a function  $y$, we can take a chain of derivatives of the function.

So we can take the first derivative

$$\frac{d}{dx}y \quad, \quad y'$$

and then the second derivative: 

$$\frac{d^2}{dx^2}y \quad, \quad y''$$

and then the third derivative: 

$$\frac{d^3}{dx^3}y \quad, \quad y''' , \quad y^{(3)}$$


### Example 1
Given the following function, find the second derivative.
$$f(x) = 5x -3$$

```python
expr = 5 * x - 3
d = sp.Derivative(expr, x, 2)
sp.Eq(d, d.doit())
```

### Example 2
Given the following function, find the second derivative.
$$f(x) = -12x^5 +\frac{1}{2}x^4-\sqrt{1-x}$$

```python
expr = -12 * x**5 + sp.Rational(1, 2) * x**4 - sp.sqrt(1 - x)
d = sp.Derivative(expr, x, 2)
sp.Eq(d, d.doit())
```

### Example 3 
Given the following function, find the second derivative.
$$f(x) = \frac{2}{(1-4x)^2}$$

```python
expr = 2 / ((1 - 4 * x) ** 2)
d = sp.Derivative(expr, x, 2)
sp.Eq(d, d.doit().simplify())
```

### Example 4 
Given the following function, find the second derivative.
$$f(x) = \sec(x)$$

```python
expr = sp.sec(x)
d = sp.Derivative(expr, x, 2)
sp.Eq(d, d.doit())
```

### Example 5
Given the following function, find the third derivative.
$$f(x) = -4x^2 + 5$$

```python
expr = -4 * x**2 + 5
d = sp.Derivative(expr, x, 3)
sp.Eq(d, d.doit())
```
