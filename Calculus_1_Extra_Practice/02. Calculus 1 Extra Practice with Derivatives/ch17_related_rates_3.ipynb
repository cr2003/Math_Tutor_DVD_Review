```python
# Define independent variable t, x as a function of t, and derivative symbols
t = sp.Symbol("t", real=True)
x = sp.Function("x")(t)
dxdt = sp.Symbol("dxdt", real=True)
drdt = sp.Symbol("drdt", real=True)
h = sp.Symbol("h", real=True, positive=True)

# Pythagorean relation: r^2 = x^2 + 5^2 => r = sqrt(x^2 + 25)
r = sp.sqrt(x**2 + h**2)

# Differentiate rope length r with respect to time t
# Substitute derivative of x(t) with dxdt for algebraic handling
diff_r = r.diff(t).subs(sp.Derivative(x, t), dxdt)

display(Math(rf"r = {sp.latex(r)}"))
display(Math(rf"\frac{{dr}}{{dt}} = {sp.latex(diff_r)}"))
```

```python
# Define AstroPy Quantity variables using the '_qty' suffix matching the whiteboard
x_qty = 12.0 * u.m
dxdt_qty = 2.0 * (u.m / u.s)
h_qty = 5.0 * u.m

# Create the numerical function using lambdify from the sympy expression
# diff_r depends on x and dxdt
f_drdt = sp.lambdify((x, dxdt, h), diff_r, modules="numpy")

# Evaluate directly using AstroPy Quantity objects
drdt_qty = f_drdt(x_qty, dxdt_qty, h_qty)

print(f"Rope pulling speed: {drdt_qty:.4f}")
display(
    Math(
        rf"\left.\frac{{dr}}{{dt}}\right|_{{x=12\text{{ m}}}} \approx \, {drdt_qty.value: .4f} \, \frac{{\text{{m}}}}{{\text{{s}}}}"
    )
)
```

```python
# Define independent variable t, horizontal distance x as a function of t, and derivative symbols
t = sp.Symbol("t", real=True)
x = sp.Function("x")(t)
dxdt = sp.Symbol("dxdt", real=True)

# Constant height of the kite
h = sp.Symbol("h", real=True, positive=True)

# Pythagorean relation: r^2 = x^2 + h^2 => r = sqrt(x^2 + h^2)
r = sp.sqrt(x**2 + h**2)

# Differentiate string length r with respect to time t
# Substitute the derivative of x(t) with dxdt for algebraic handling
diff_r = r.diff(t).subs(sp.Derivative(x, t), dxdt)

display(Math(rf"r = {sp.latex(r)}"))
display(Math(rf"\frac{{dr}}{{dt}} = {sp.latex(diff_r)}"))
```

```python
# Define AstroPy Quantity variables matching the whiteboard and problem statement
h_qty = 100.0 * u.m
r_qty = 200.0 * u.m
dxdt_qty = 10.0 * (u.m / u.s)

# Calculate corresponding horizontal distance x at r = 200 m using NumPy/AstroPy
x_qty = np.sqrt(r_qty**2 - h_qty**2)

# Create the numerical function using lambdify from the sympy expression
# diff_r depends on x, dxdt, and h
f_drdt = sp.lambdify((x, dxdt, h), diff_r, "numpy")

# Evaluate directly using AstroPy Quantity objects
drdt_qty = f_drdt(x_qty, dxdt_qty, h_qty)

print(f"Rate to let out the string: {drdt_qty:.4f}")
display(
    Math(
        rf"\left.\frac{{dr}}{{dt}}\right|_{{r=200\text{{ m}}}} \approx \, {drdt_qty.value: .4f} \, \frac{{\text{{m}}}}{{\text{{s}}}}"
    )
)
```

```python
# Define independent variable t, angle theta as a function of t, and derivative symbol
t = sp.Symbol("t", real=True)
theta = sp.Function("theta")(t)
dydt = sp.Symbol("dydt", real=True)

# Define variables for rocket height y(t) and constant ground distance x
y = sp.Function("y")(t)
x = sp.Symbol("x", real=True, positive=True)

# Trigonometric relation from the right triangle: tan(theta) = y / x => theta = atan(y / x)
theta_expr = sp.atan(y / x)

# Differentiate theta with respect to time t
# Substitute the derivative of y(t) with dydt for algebraic handling
diff_theta = theta_expr.diff(t).subs(sp.Derivative(y, t), dydt)

display(Math(rf"\theta = {sp.latex(theta_expr)}"))
display(Math(rf"\frac{{d\theta}}{{dt}} = {sp.latex(diff_theta)}"))
```

```python
# Define AstroPy Quantity variables matching the whiteboard and problem statement
y_qty = 2.0 * u.km
x_qty = 5.0 * u.km
dydt_qty = 300.0 * (u.km / u.h)

# Create the numerical function using lambdify from the sympy expression
# diff_theta depends on y, dydt, and x
f_dthetadt = sp.lambdify((y, dydt, x), diff_theta, "numpy")

# Evaluate directly using AstroPy Quantity objects
dthetadt_qty = f_dthetadt(y_qty, dydt_qty, x_qty) * (u.rad / u.h)

print(f"Rate of change of elevation angle: {dthetadt_qty:.4f}")
display(
    Math(
        rf"\left.\frac{{d\theta}}{{dt}}\right|_{{y=2\text{{ km}}}} \approx \, {dthetadt_qty.value: .4f} \, \frac{{\text{{rad}}}}{{\text{{h}}}}"
    )
)
```
