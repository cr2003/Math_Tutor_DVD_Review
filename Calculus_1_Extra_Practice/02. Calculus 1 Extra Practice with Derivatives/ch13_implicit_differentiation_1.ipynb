## Differentitation of an imiplicit equation

Implicit differentiation is when we don't have a explicit equality  of the variable, (solved by y), such as:

$$3y^2 = -2x^4$$

then we differentiate both sides of the equation: 

$$6y \frac{dy}{dx} = -8x^3$$

and solved by de $\frac{dy}{dx}$: 

$$\frac{dy}{dx} = \frac{-4x^3}{3y}$$


### Example 1
Given the following equation 

$$3y^2 = -2x^4$$

find $\frac{dy}{dx}$ using implicit differentiation.


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(3 * y**2, -2 * x**4)
expr = 3 * y**2 + 2 * x**4
dy_dx = sp.idiff(expr, y, x)
d = dy / dx
eq = sp.Eq(d, dy_dx)
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
```

### Example 2
Given the following equation 

$$y^2 + y = \frac{1+x}{1-x}$$

find $\frac{dy}{dx}$ using implicit differentiation.


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(y**2 + y, (1 + x) / (1 - x))
expr = y**2 + y - ((1 + x) / (1 - x))
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
```

### Example 3
Given the following equation 

$$\sec y - \tan x = 0$$

find $\frac{dy}{dx}$ using implicit differentiation.


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(sp.sec(y) - sp.tan(x), 0)
expr = sp.sec(y) - sp.tan(x)
dy_dx = sp.idiff(expr, y, x)
d = dy / dx
eq = sp.Eq(d, dy_dx)
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
```

### Example 4
Given the following equation 

$$\frac{\sin(y)}{y^2+1} = 3x$$

find $\frac{dy}{dx}$ using implicit differentiation.


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(sp.sin(y) / (y**2 + 1), 3 * x)
expr = sp.sin(y) / (y**2 + 1) - 3 * x
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
```

### Example 5
Given the following equation 

$$x^2 + x^2y^2 + y^3 = 3$$

find $\frac{dy}{dx}$ using implicit differentiation.


```python
dy, dx = sp.symbols("dy, dx", real=True)
equation = sp.Eq(x**2 + x**2 * y**2 + y**3, 3)
expr = x**2 + x**2 * y**2 + y**3 - 3
dy_dx = sp.idiff(expr, y, x).factor()
d = dy / dx
eq = sp.Eq(d, dy_dx)
display(Math(rf"{sp.latex(equation)} \implies {sp.latex(eq)}"))
```
